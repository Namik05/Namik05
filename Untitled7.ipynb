```python
import time
import random
import json
import paho.mqtt.client as mqtt

# Configuration HiveMQ Cloud
BROKER = "512ea1b6209f4558ad76043ad9adaa08.s1.eu.hivemq.cloud"
PORT = 8883
USERNAME = "Abdullah"
PASSWORD = "Harrougui@2002"
TOPIC = "iot/stock"

# Seuil de sécurité
SEUIL_SECURITE = 20

# Stock initial
stock = 50

# Callback connexion
def on_connect(client, userdata, flags, rc):
    if rc == 0:
        print("✅ Connecté au broker HiveMQ Cloud.")
    else:
        print(f"❌ Erreur de connexion. Code : {rc}")

# Configuration client MQTT
client = mqtt.Client()
client.username_pw_set(USERNAME, PASSWORD)
client.tls_set()
client.on_connect = on_connect
client.connect(BROKER, PORT)

# Boucle de simulation
try:
    while stock > SEUIL_SECURITE:
        consommation = random.randint(1, 5)  # Consommation aléatoire
        stock -= consommation
        stock = max(stock, 0)  # Évite les valeurs négatives

        payload = json.dumps({"stock_actuel": stock})
        client.publish(TOPIC, payload)
        print(f"📦 Stock actuel : {stock} pièces (consommé : {consommation})")

        time.sleep(3)

    print(f"⚠️  Stock critique atteint : {stock} pièces. Simulation arrêtée.")
    client.disconnect()

except KeyboardInterrupt:
    print("🛑 Simulation arrêtée manuellement.")
    client.disconnect()
```
